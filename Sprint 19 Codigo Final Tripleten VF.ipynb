***Métricas***:
Interconnect quiere: detectar clientes que probablemente se vayan para ofrecerles una acción de retención.   
Por lo tanto, un falso negativo puede ser costoso debido a:   

`Cliente iba a cancelar -> Modelo dice "no cancelará" -> No recibe promoción -> Interconnect pierde cliente`   

Por eso debemos prestar especial atención a: ***Recall***

Reflexión "De todos los clientes que realmente cancelaron, ¿cuántos conseguimos detectar?, el maximizar ***Recall*** indiscriminadamente, porque terminaríamos ofreciendo promociones a demasiados clientes que nunca iban a cancelar". aplicacar lo anterior tambien cuesta y mucho!!!
Personalmente me inclino a comparar las siguientes métricas:   

***ROC-AUC + Recall + Precision + F1*** y analizar la matriz de confusión.

Para Interconnect, el FN (Falso Negativo) es particularmente importante.   

| Real         | Predicción   | Resultado |
|--------------|--------------|-----------|
| No abandona  | No abandona  | ✅ TN     |
| No abandona  | Abandona     | ❌ FP     |
| Abandona     | No abandona  | ⚠️ FN     |
| Abandona     | Abandona     | ✅ TP     |



***FN*** = clientes que realmente abandonaron, pero el modelo dijo que no abandonarían.
Un FN puede ser costoso, porque significa que no se detectó a un cliente que estaba en riesgo de abandono.

Si el objetivo es detectar la mayor cantidad posible de clientes en riesgo, se debe prestar especial atención al Recall (sensibilidad) y si se desea un euilibrio con presición entonces la métrica `F1-score` es mejor porque busca un equilibrio entre detectar clientes que realmente abandonarán ***(Recall)*** y no señalar innecesariamente a clientes que no abandonarán ***(Precision)***.

* 🎯 F1 → mejor equilibrio.
* 🔎 Recall → mejor para detectar la mayor cantidad de abandonos.
* 🎯 Precision → mejor para que las ofertas sean más acertadas.
* 📊 AUC-ROC → mejor para comparar la capacidad general de los modelos.

## Prueba de Hipótesis

***H₀ (hipótesis nula):***

El tipo de contrato y el abandono del cliente son independientes.
Es decir, el tipo de contrato no está relacionado con el churn.

***H₁ (hipótesis alternativa):***

El tipo de contrato y el abandono del cliente no son independientes.
Es decir, existe una relación entre el tipo de contrato y el churn.

````python
import numpy as np
from scipy.stats import chi2_contingency

tabla = pd.crosstab(df['Type'], df['churn'])
chi2, p, gl, esperados = chi2_contingency(tabla)

n = tabla.sum().sum()
r, k = tabla.shape

cramers_v = np.sqrt(
    chi2 / (n * min(r - 1, k - 1))
)

print("Chi²:", chi2)
print("p-value:", p)
print("Cramér's V:", cramers_v)
print("Grados de libertad:", gl)
````

````python
alpha = 0.05

if p < alpha:
    print("Rechazamos H0")
else:
    print("No rechazamos H0")
````

***Análisis:***

Se construyó una tabla de contingencia y se aplicó una prueba Chi-cuadrada de independencia.

***Resultado descriptivo:***   
Se rechaza Ho y se acepta Ha: El tipo de contrato y el abandono del cliente no son independientes.   
Es decir, existe una relación entre el tipo de contrato y el churn.   

El churn es de aproximadamente ***42.7 %*** en clientes Month-to-month, frente a ***11.3 %*** en contratos One year y ***2.8 %*** en Two year.

***Interpretación:***

Los clientes con contratos mensuales presentan una proporción de abandono considerablemente mayor que aquellos con contratos de uno o dos años.

***Implicación para Interconnect:***

Los clientes Month-to-month representan un segmento prioritario para estrategias de retención, como promociones, incentivos para migrar a contratos de mayor duración o beneficios exclusivos.

***Conclusión:*** La prueba Chi-cuadrada mostró una relación estadísticamente significativa entre el tipo de contrato y el churn (χ² = 1184.60, p < 0.05). Además, el valor de Cramér’s V = 0.410 indica una asociación moderada entre ambas variables. Los clientes con contratos Month-to-month presentan la mayor tasa de abandono (42.7 %), muy superior a la observada en contratos One year (11.3 %) y Two year (2.8 %).

La pregunta ahora es

Dentro de cada tipo de contrato, ¿los clientes que abandonan tienen cargos mensuales diferentes a los clientes que permanecen?

Haremos t de Welch para muestras independientes, porque no necesitamos asumir que las varianzas sean iguales.

````python
from scipy.stats import ttest_ind

tipos = df_anova['Type'].unique()

for tipo in tipos:
    
    activos = df_anova[
        (df_anova['Type'] == tipo) &
        (df_anova['churn'] == 0)
    ]['MonthlyCharges']
    
    abandonos = df_anova[
        (df_anova['Type'] == tipo) &
        (df_anova['churn'] == 1)
    ]['MonthlyCharges']
    
    t, p = ttest_ind(
        activos,
        abandonos,
        equal_var=False
    )
    
    print(f'{tipo}')
    print(f'  t = {t:.3f}')
    print(f'  p-value = {p:.6g}')
    print()
````

Pero hagámoslo correctamente con corrección por múltiples comparaciones

Estamos haciendo 3 pruebas simultáneas:

1. Month-to-month
2. One year
3. Two year

Por lo tanto, es recomendable ajustar los p-values. Una opción sencilla es Bonferroni:   
Usaremos:

\alpha = 0.05   

p_ajustado < 0.05  → diferencia significativa   
p_ajustado ≥ 0.05  → no significativa

````python
from scipy.stats import ttest_ind
from statsmodels.stats.multitest import multipletests

resultados = []

for tipo in df_anova['Type'].unique():
    
    activos = df_anova[
        (df_anova['Type'] == tipo) &
        (df_anova['churn'] == 0)
    ]['MonthlyCharges']
    
    abandonos = df_anova[
        (df_anova['Type'] == tipo) &
        (df_anova['churn'] == 1)
    ]['MonthlyCharges']
    
    t, p = ttest_ind(
        activos,
        abandonos,
        equal_var=False
    )
    
    resultados.append({
        'Type': tipo,
        't': t,
        'p_value': p
    })

resultados = pd.DataFrame(resultados)

resultados['p_ajustado'] = multipletests(
    resultados['p_value'],
    method='bonferroni'
)[1]

resultados
````

Pero no debemos concluir solamente por la diferencia de medias. 
Ahora aplicando la prueba ***Bonferroni*** y checando sus resultados:
```

0	Month-to-month	-13.810993	2.287778e-42	6.863335e-42

```

Se determina que las diferencias obtenidas en cada grupo ***son estadísticamente significativas***.

Dentro de cada tipo de contrato, los clientes que abandonaron presentan cargos mensuales significativamente mayores que los clientes que permanecieron ademas se observa que la diferencia es mayor en los contratos de mayor duración.

***Conclusión:*** La ***ANOVA*** de dos factores mostró diferencias significativas en MonthlyCharges asociadas al tipo de contrato y al estado de churn. Además, se encontró una interacción significativa entre ambas variables (p = 1.116 × 10⁻⁶), indicando que la diferencia de cargos mensuales entre clientes activos y clientes que abandonan depende del tipo de contrato.

Los análisis de efectos simples confirmaron diferencias significativas entre `activos y abandonos` en los tres tipos de contrato, incluso después de aplicar la corrección de Bonferroni. Los clientes que abandonaron presentaron cargos mensuales promedio superiores en todos los casos: $73.02 vs $61.46 en Month-to-month, $85.05 vs $62.51 en One year y $86.78 vs $60.01 en Two year.

***Implicación para el negocio:*** los cargos mensuales elevados podrían constituir una ***señal relevante*** para identificar clientes con mayor riesgo de abandono, pero su efecto debe analizarse conjuntamente con el tipo de contrato.

````python
from scipy.stats import ttest_ind

activos = df.loc[
    df['churn'] == 0,
    'tenure_days'
].dropna()

abandonos = df.loc[
    df['churn'] == 1,
    'tenure_days'
].dropna()

t, p = ttest_ind(
    activos,
    abandonos,
    equal_var=False
)

print(f't = {t:.4f}')
print(f'p-value = {p:.6g}')
````

````python
import numpy as np

activos = df.loc[
    df['churn'] == 0,
    'tenure_days'
].dropna()

abandonos = df.loc[
    df['churn'] == 1,
    'tenure_days'
].dropna()

d = (
    activos.mean() - abandonos.mean()
) / np.sqrt(
    (
        (len(activos)-1) * activos.var() +
        (len(abandonos)-1) * abandonos.var()
    ) /
    (len(activos) + len(abandonos) - 2)
)

print(f"Cohen's d = {d:.4f}")
````

La antigüedad del cliente presenta una diferencia estadísticamente significativa entre los clientes activos y los que abandonaron (Welch t-test, t = 30.32, p < 0.001). El tamaño del efecto fue de Cohen's d = 0.7429, lo que indica una diferencia de magnitud moderada-alta.

Estos resultados sugieren que la antigüedad es una variable relevante para comprender el comportamiento de churn, ya que los clientes con menor permanencia presentan una mayor propensión al abandono.

| `tenure_days → churn` | Cohen's d | **0.7429** | Efecto moderado-alto |
| `Fiber optic: tenure → churn` | Cohen's d | **0.8839** | Efecto grande |

### Principales hallazgos

- El **tipo de contrato** presenta una asociación moderada con el churn (`V = 0.410`).
- El **tipo de servicio de Internet** también presenta una asociación moderada con el churn (`V = 0.323`).
- Los **cargos mensuales** presentan una diferencia estadísticamente significativa respecto al churn, pero su tamaño de efecto es pequeño (`η²p = 0.0334`).
- La **antigüedad del cliente** presenta un efecto considerablemente mayor (`d = 0.7429`).
- Dentro del segmento **Fiber optic**, la diferencia de antigüedad entre activos y abandonos alcanza un **efecto grande** (`d = 0.8839`).

### Conclusión general

Los resultados sugieren que la **antigüedad del cliente** es una de las variables más relevantes para comprender el comportamiento de churn. Este efecto es todavía más marcado dentro del segmento de clientes con **Fiber optic**.

Por otro lado, el tipo de contrato y el servicio de Internet presentan asociaciones moderadas con el abandono, mientras que los cargos mensuales, aunque estadísticamente significativos, muestran un efecto relativamente pequeño.

Estos hallazgos proporcionan una base para seleccionar y analizar las variables en la siguiente etapa de **Machine Learning**, cuyo objetivo será predecir la probabilidad de abandono de cada cliente.

````python
# Optimización de Thresholds
from sklearn.metrics import precision_recall_curve

precision_rf, recall_rf, thresholds_rf = precision_recall_curve(
    y_val,
    y_prob_rf_val
)

# Calculo de F1
f1_rf = (
    2 * precision_rf[:-1] * recall_rf[:-1]
    / (precision_rf[:-1] + recall_rf[:-1] + 1e-10)
)

# Indice de maximi F1
indice_mejor_rf = np.argmax(f1_rf)

# Threshold optimizada
threshold_optimo_rf = thresholds_rf[indice_mejor_rf]

# Resultados
print("Threshold óptimo rf:", threshold_optimo_rf)
print("Precision:", precision_rf[indice_mejor_rf])
print("Recall:", recall_rf[indice_mejor_rf])
print("F1:", f1_rf[indice_mejor_rf])
print("ROC-AUC   :", roc_auc_score(y_val, y_prob_rf_val))
````

Si Interconnect pierde mucho dinero cuando un cliente abandona y el costo de ofrecer una promoción preventiva es relativamente bajo, podría ser preferible el ***Random Forest con threshold 0.2733***, aunque genere más falsos positivos.

Si las ***campañas de retención son costosas***, podríamos preferir el ***Decision Tree***, porque su Precision es considerablemente mayor.

Para ***Interconnect***, esta diferencia debe analizarse desde una perspectiva de negocio. Un modelo con mayor recall puede ser conveniente si el costo de no detectar a un cliente que finalmente abandona ***(FN)*** es considerablemente mayor que el costo de contactar a un cliente que finalmente permanece ***(FP)***. Por el contrario, si las campañas de retención tienen un costo elevado, también es importante controlar los falsos positivos para evitar ofrecer promociones innecesariamente.

Por ello, el threshold seleccionado mediante la maximización del F1-score no necesariamente representa el punto óptimo desde una perspectiva económica. Se requiere realizar un análisis costo-beneficio de los falsos positivos ***(FP)*** y falsos negativos ***(FN)*** para determinar el threshold que maximice el beneficio esperado para Interconnect.

En consecuencia, F1-score es una métrica adecuada para evaluar el equilibrio entre precision y recall durante la selección inicial del modelo, pero la decisión final del threshold debe basarse en los costos y beneficios asociados a cada tipo de error. 

Si el costo de perder un cliente es considerablemente mayor que el costo de una campaña de retención, podría ser conveniente seleccionar un threshold más bajo para incrementar el recall, aun cuando esto genere un mayor número de falsos positivos.
